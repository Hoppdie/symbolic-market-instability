# Data Exploration

This notebook explores the downloaded market data, plots OHLCV and VIX time series, shows basic statistics, and identifies crash periods visually.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Add src to path
project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))
sys.path.insert(0, str(project_root / "src"))

from src.data_acquisition.downloader import MarketDataDownloader
from src.data_acquisition.preprocessor import DataPreprocessor
```

```python
# Download sample data
downloader = MarketDataDownloader()
data = downloader.get_combined_data("^GSPC", "2000-01-01", "2024-12-31", "US")

print(f"Downloaded {len(data)} days of data")
print(f"Date range: {data.index.min()} to {data.index.max()}")
data.head()
```

```python
# Basic statistics
data.describe()
```

```python
# Plot OHLCV data
fig, axes = plt.subplots(3, 1, figsize=(14, 10))

# Close price
axes[0].plot(data.index, data['Close'], linewidth=1.5, color='blue')
axes[0].set_ylabel('Close Price', fontsize=12)
axes[0].set_title('S&P 500 Close Price', fontsize=14, fontweight='bold')
axes[0].grid(True, alpha=0.3)

# Volume
axes[1].plot(data.index, data['Volume'], linewidth=1, color='green', alpha=0.7)
axes[1].set_ylabel('Volume', fontsize=12)
axes[1].set_title('Trading Volume', fontsize=14, fontweight='bold')
axes[1].grid(True, alpha=0.3)

# VIX
axes[2].plot(data.index, data['VIX'], linewidth=1.5, color='red', alpha=0.7)
axes[2].set_ylabel('VIX', fontsize=12)
axes[2].set_xlabel('Date', fontsize=12)
axes[2].set_title('Volatility Index (VIX)', fontsize=14, fontweight='bold')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('../results/figures/data_exploration.png', dpi=300, bbox_inches='tight')
plt.show()
```

# Case Study: 2008 Financial Crisis

This notebook focuses on the 2007-2009 period, shows symbolization during the crisis, displays fired rules timeline, and analyzes lead time.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Add src to path
project_root = Path().resolve().parent
sys.path.insert(0, str(project_root))
sys.path.insert(0, str(project_root / "src"))

from src.evaluation.case_studies import CaseStudyEvaluator
```

```python
# Load analysis results
results_file = project_root / "results" / "analysis_results.csv"
if results_file.exists():
    results = pd.read_csv(results_file, index_col=0, parse_dates=True)
    print(f"Loaded {len(results)} days of results")
    
    # Filter to 2007-2009 period
    crisis_period = results[
        (results.index >= '2007-01-01') & 
        (results.index <= '2009-12-31')
    ]
    print(f"Crisis period data: {len(crisis_period)} days")
else:
    print("Analysis results not found. Please run scripts/run_analysis.py first.")
```

```python
# Evaluate 2008 crisis
evaluator = CaseStudyEvaluator()
metrics = evaluator.evaluate_2008_crisis(results)

print("2008 Crisis Evaluation:")
for key, value in metrics.items():
    print(f"  {key}: {value}")
```

```python
# Plot timeline for crisis period
if 'crisis_period' in locals() and not crisis_period.empty:
    evaluator.plot_timeline(crisis_period, ['2008-09-15'])
    print("Timeline plot saved to results/figures/instability_timeline.png")
```
